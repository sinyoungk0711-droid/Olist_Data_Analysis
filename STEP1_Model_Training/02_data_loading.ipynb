````python
# 셀 2. 라이브러리 불러오기

import os       # 폴더 생성, 파일 목록 확인
import zipfile  # zip 파일 압축 해제
import pandas as pd  # 표 형태 데이터(DataFrame) 처리
````

````python
# 셀 3. 데이터 저장 폴더 설정
# 앞으로 모든 PART에서 동일하게 사용할 경로입니다.

base_path = "/content/drive/MyDrive/Colab Notebooks/Data_Analysis/Olist/data"

os.makedirs(base_path, exist_ok=True)
# exist_ok=True → 폴더가 이미 있어도 오류가 나지 않습니다.

print("데이터 저장 위치:", base_path)
````

````python
# 셀 4. Kaggle에서 데이터 다운로드
# kagglehub: Kaggle 데이터셋을 코드 한 줄로 내려받는 도구
# 받은 파일은 Colab 임시 폴더(path)에 압축이 풀린 상태로 저장됩니다.

import kagglehub

# Download latest version
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")

print("Path to dataset files:", path)
````

````python
# 셀 5. 다운로드한 파일 확인
# 임시 폴더(path)에 CSV 파일이 몇 개 있는지 확인합니다.

downloaded = sorted(f for f in os.listdir(path) if f.endswith(".csv"))
print(f"다운로드한 CSV: {len(downloaded)}개")
# ▶ 9개면 정상입니다.
````

````python
# 셀 6. Drive 폴더로 복사
# 임시 폴더는 Colab을 끄면 사라지므로, 모든 PART에서 쓰도록 Drive(base_path)로 옮겨 둡니다.

import shutil  # 파일 복사
for name in downloaded:
    shutil.copy(os.path.join(path, name), base_path)

print("복사 완료!")
print("저장 위치:", base_path)
````

````python
# 셀 7. CSV 파일 목록 확인
# 압축이 풀린 CSV 파일 목록을 확인합니다.

csv_files = [f for f in os.listdir(base_path) if f.endswith(".csv")]

print(f"CSV 파일 개수: {len(csv_files)}개")
print()
for name in sorted(csv_files):
    print(" -", name)
# ▶ 처음 실행하면 9개 파일이 보이면 정상입니다.
#   (PART 4를 끝낸 뒤 다시 실행하면 olist_master_data.csv가 더해져 10개가 보입니다.)
````

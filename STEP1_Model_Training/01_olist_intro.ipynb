## 1. Olist란?

브라질의 실제 이커머스 플랫폼입니다.
Kaggle에서 공개한 오픈 데이터이며, 2016~2018년 약 10만 건의 주문 정보가 담겨 있습니다.

| 항목 | 내용 |
|---|---|
| 데이터 이름 | Olist E-commerce Dataset |
| 출처 | Kaggle |
| 국가 | 브라질 |
| 기간 | 2016 ~ 2018 |
| 주문 수 | 약 10만 건 |


## 2. 왜 이 데이터를 쓰나요?

| 이유 | 설명 |
|---|---|
| 실무형 구조 | 실제 쇼핑몰과 비슷한 데이터 구조 |
| 다양한 AI 실습 | 유사 상품 추천(KNN), 리뷰 감성 분석(머신러닝·딥러닝·BERT), 판매자 군집(비지도 학습), 판매자 이탈 예측(분류) 가능 |
| 관계형 데이터 경험 | 여러 CSV를 연결(merge)하는 실습 가능 |


## 3. 포함된 CSV 파일

| 파일명 | 설명 |
|---|---|
| customers | 고객 정보 |
| orders | 주문 정보 |
| order_items | 주문 상품 정보 |
| products | 상품 정보 |
| payments | 결제 정보 |
| reviews | 리뷰 정보 |
| sellers | 판매자 정보 |
| geolocation | 위치 정보 |


## 4. 수업 전체 흐름

```
PART 1    → 데이터셋 소개
PART 2    → 데이터 다운로드 및 저장
PART 3    → 데이터 구조 이해
PART 4    → 데이터 병합(merge)
PART 5    → EDA (탐색적 데이터 분석)
PART 6    → 유사 상품 추천 (규칙 기반 vs KNN)
PART 7-1  → 리뷰 감성 분석 ① 머신러닝 (TF-IDF + 로지스틱 회귀)
PART 7-2  → 리뷰 감성 분석 ② 딥러닝 (Embedding · LSTM)
PART 7-3  → 리뷰 감성 분석 ③ 사전학습 모델 (BERT)
PART 8    → 판매자 유형 나누기 (비지도 학습 · K-Means)
PART 9    → 판매자 이탈 예측 (로지스틱 회귀 vs 랜덤 포레스트)
PART 10   → 비즈니스 전략
```

## 5. 핵심 정리

1. 데이터는 분석과 AI의 재료입니다.
2. Olist 데이터는 실제 쇼핑몰 구조를 경험하기 좋습니다.
3. 여러 데이터를 연결해야 의미 있는 분석이 가능합니다.
4. AI는 정답을 주는 도구가 아니라 더 나은 판단을 돕는 도구입니다.
